```python
#Pipe;line for scilearn

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction import text

from sklearn.naive_bayes import GaussianNB

from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer
```

```python
df = pd.read_csv('cleaned_reviews.csv')

df
```

```python
X = df["text"]

Y = df["label"]
```

```python
def summarize_classification(y_test, y_pred):

    acc = accuracy_score(y_test, y_pred, normalize=True)
    num_acc = accuracy_score(y_test, y_pred, normalize=False)
    prec = precision_score(y_test, y_pred, average='weighted')
    recall = recall_score(y_test, y_pred, average='weighted')

    print("Length of testing data: ", len(y_test))
    print("accuracy_count : " , num_acc)
    print("accuracy_score : " , acc)
    print("precision_score : " , prec)
    print("recall_score : ", recall)
```

```python
pipeline = Pipeline([
    ('vectorizer', CountVectorizer(ngram_range=(2, 2))),
    ('to_dense', FunctionTransformer(lambda x: np.asarray(x.todense()), validate=False)),
    ('clf', GaussianNB())
])

pipeline
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
pipeline.fit(x_train, y_train)

y_pred = pipeline.predict(x_test)

y_pred
```

```python
summarize_classification(y_test, y_pred)
```

```python
def text_classification_with_ngrams(X, Y, ngram_ranges):
    for ngram_range in ngram_ranges:

        pipeline = Pipeline([
            ('vectorizer', CountVectorizer(ngram_range=ngram_range)),
            ('to_dense', FunctionTransformer(lambda x: np.asarray(x.todense()), validate=False)),
            ('clf', GaussianNB())
        ])

        X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

        pipeline.fit(X_train, Y_train)

        y_pred = pipeline.predict(X_test)

        print(f"Classification Report for ngram_range={ngram_range}:\n")
        summarize_classification(Y_test, y_pred)
        print('-'*80)
```

```python
text_classification_with_ngrams(X, Y, ngram_ranges=[(1, 1), (1, 2), (1, 3), (4, 4)])
```
